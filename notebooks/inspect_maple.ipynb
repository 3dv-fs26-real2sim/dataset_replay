```python
import h5py
import numpy as np
import pandas as pd

# Open and inspect the H5 file
h5_path = "/home/shinben0327/3dv/dataset_replay/data/h5/20250922_143954.h5"

with h5py.File(h5_path, 'r') as f:
    print("=" * 60)
    print("H5 File Structure")
    print("=" * 60)
    
    def print_structure(name, obj):
        indent = "  " * name.count('/')
        if isinstance(obj, h5py.Dataset):
            print(f"{indent}{name}: shape={obj.shape}, dtype={obj.dtype}")
        else:
            print(f"{indent}{name}/ (Group)")
    
    f.visititems(print_structure)
    
    print("\n" + "=" * 60)
    print("Camera Intrinsics")
    print("=" * 60)
    if 'observations/images/oakd_front_view/intrinsics' in f:
        intrinsics = f['observations/images/oakd_front_view/intrinsics'][:]
        print("Intrinsic Matrix (K):")
        print(intrinsics.reshape(3, 3))
    
    print("\n" + "=" * 60)
    print("Camera Extrinsics")
    print("=" * 60)
    if 'observations/images/oakd_front_view/extrinsics' in f:
        extrinsics = f['observations/images/oakd_front_view/extrinsics'][:]
        print("Extrinsic Matrix (4x4):")
        print(extrinsics.reshape(4, 4))
    
    print("\n" + "=" * 60)
    print("Dataset Summary")
    print("=" * 60)
    print(f"Total groups/datasets: {len(list(f.keys()))}")
    for key in f.keys():
        if isinstance(f[key], h5py.Dataset):
            print(f"{key}: {f[key].shape}")
```
